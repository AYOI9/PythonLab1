```python
import random
```

```python
arr = [5, 15] + random.sample(range(16, 100), 8)
random.shuffle(arr)
print(f"Исходный массив: {arr}")

idx5 = arr.index(5)
idx15 = arr.index(15)
arr[idx5], arr[idx15] = arr[idx15], arr[idx5]

print(f"Результат: {arr}")
```

```python
n = 4
nums = random.sample(range(1, 100), n * n)
matrix = [nums[i*n:(i+1)*n] for i in range(n)]

print("Матрица:")
for row in matrix: print(row)

max_val = -1
min_val = 101
max_row_idx = 0
min_col_idx = 0

for r in range(n):
    for c in range(n):
        if matrix[r][c] > max_val:
            max_val = matrix[r][c]
            max_row_idx = r
        if matrix[r][c] < min_val:
            min_val = matrix[r][c]
            min_col_idx = c

row_vec = matrix[max_row_idx]
col_vec = [matrix[i][min_col_idx] for i in range(n)]
scalar_product = sum(row_vec[i] * col_vec[i] for i in range(n))

print(f"Строка с MAX ({max_val}): {row_vec}")
print(f"Столбец с MIN ({min_val}): {col_vec}")
print(f"Скалярное произведение: {scalar_product}")
```

```python
nums = [random.randint(15, 25) for _ in range(15)]
print(f"Исходный список: {nums}")

if 20 in nums:
    idx = nums.index(20)
    nums[idx] = 200

print(f"Результат: {nums}")
```

```python
list1 = [random.randint(1, 15000) for _ in range(10000)]
list2 = [random.randint(1, 15000) for _ in range(10000)]

result = sorted(list(set(list1) & set(list2)))

print(f"Количество общих чисел: {len(result)}")
print(f"Результат (первые 50 чисел): {result[:50]}")
```
